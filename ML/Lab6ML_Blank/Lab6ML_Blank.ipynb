# SVR

# Importing the libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

# Read World bank CSV

```python
df = pd.read_csv("WorldBank.csv")
```

```python
df
```

# Perform conditional selection to find - Population ages 15-64 (% of total population)

```python
data = df[df["Indicator Name"] == "Population ages 15-64 (% of total population)"]
```

```python
data
```

# Divide the data into input and output

```python
X =np.arange(1960,2020).reshape(-1,1)
```

```python
X
```

```python
data.values
```

```python
y = data.values[0][4:-1].reshape(-1,1)
```

```python
y
```

# Plot scatter plot of Population ages 15-64 (% of total population)

```python
plt.scatter(X,y)
```

# Feature Scaling (Mandatory for SVR)**
SVR is highly sensitive to the range of data points. If we don't scale (normalize) the data, the model will fail to find the correct hyperplane.

```python
from sklearn.preprocessing import StandardScaler

sc_X = StandardScaler()
sc_y = StandardScaler()
X_scaled = sc_X.fit_transform(X)
y_scaled = sc_y.fit_transform(y)

print("Scalling Complete")
```

```python
y_scaled
```

# Fitting SVR on Linear Kernel on dataset

```python
from sklearn.svm import SVR

svr_lin = SVR(kernel="linear")
y_scaled
```

```python
X_scaled
```

```python
svr_lin.fit(X_scaled,y_scaled.ravel())
```

# Predict Values using Model

```python
X_predicted = svr_lin.predict(X_scaled)
X_predicted
```

```python
# use inverse_transform
y_predicted_lin = sc_y.inverse_transform(X_predicted.reshape(-1,1))
```

```python
y_predicted_lin
```

```python
plt.scatter(X,y)
plt.scatter(X,y_predicted_lin)
```

# Predict the X_scaled using 3 Kernel

```python
# above Steps Only - Kernel Change
svr_poly = SVR(kernel="poly",degree=3)
svr_rbf = SVR(kernel ="rbf")
```

```python
svr_poly.fit(X_scaled,y_scaled.ravel())
svr_rbf.fit(X_scaled,y_scaled.ravel())
```

# Visualising the  results

```python
plt.figure(figsize=(12,8))
plt.scatter(X,y, color='darkorange', label='data')
plt.plot(X, y_predicted_lin, color='navy' , label='Linear model')
y_predicted_rbf = sc_y.inverse_transform(svr_rbf.predict(X_scaled).reshape(-1,1))
plt.plot(X,y_predicted_rbf,color='cornflowerblue',label="polynomial model")
y_predicted_rbf = sc_y.inverse_transform(svr_rbf.predict(X_scaled).reshape(-1,1))
plt.plot(X,y_predicted_rbf,color='green',label="RBF model")
plt.xlabel('data')
plt.ylabel('target')
plt.title('support vector regression : kernel Comparison')
plt.legend()
plt.show()
```
