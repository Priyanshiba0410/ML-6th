# Decision Tree Classifier

```python
from sklearn.datasets import load_iris 
```

# Import iris data ser using sklearn

```python
data = load_iris()
```

```python
data
```

# Importing the libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn import tree
```

# Divide the data into input and output

```python
X = data['data']
y = data['target']
y
```

# Splitting the dataset into the Training set and Test set

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3
)
```

```python
X_test
```

# Fitting DecisionTreeClassifier   on dataset

```python
model = DecisionTreeClassifier()
```

```python
model.fit(X_train,y_train)
```

# Display Decision Tree

```python
%matplotlib inline
```

```python
tree.plot_tree(model)
```

```python
plt.show()
```

# Predict the x_test 

```python
model.predict(X_test)
```

```python
y_test
```

```python
model.score(X_train,y_train)
```

```python
model.score(X_test,y_test)
```

# Import diabetes.csv dataset 

```python
df = pd.read_csv("diabetes.csv")
df
```

# Check the distribution of the target

```python
print(df['Outcome'].value_counts())
```

# Replace zeros with NaN for specific columns

```python
columns_to_replace = ["Pregnancies","Glucose","BloodPressure","SkinThickness","Insulin","BMI"]
df[columns_to_replace] = df[columns_to_replace].replace(0,np.nan)
```

# Check for missing values

```python
print(df.isnull().sum())
```

# Fill missing values with median

```python
for col in columns_to_replace:
    df[col] = df[col].fillna(df[col].median())
```

```python
print(df.isnull().sum())
```

# Visualize Distributions

```python
import seaborn as sns
sns.pairplot(df,hue='Outcome')
```

# Convert Targer data into interger code

```python
df['Outcome'] = df['Outcome'].map({"Diabetic":1,"Non-diabetic":0})
```

```python
sns.heatmap(df.corr(),anno = True ,cmap='coolwarm')
plt.show()
```
